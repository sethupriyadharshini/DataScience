# FREQUENCY

```python
all_freq_tables = {} # Dictionary to store results

for columnName in quan:
    all_freq_tables[columnName]=Univariate.freqTable(columnName,dataset)
```

```python
# Example: Access the frequency table for 'ssc_p'
all_freq_tables['ssc_p']
```

```python
#Print all quantity frequency table details
for col_name, table in all_freq_tables.items():
    print(f"Frequency table for: {col_name}")
    print(table)
    print("-" * 40)
```
